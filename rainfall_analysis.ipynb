# Rainfall Analysis and Planting Advisory

**Project:** PLP Data Analysis Assignment 2

**Objective:** Analyse monthly rainfall patterns across five Kenyan counties using Python and NumPy to identify suitable planting windows and provide practical agricultural advice.

```python
import numpy as np
import csv
```

```python
counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]

months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

clean_rows = [
    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],
    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],
    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],
    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70]
]

machakos_raw = [
    "48", "35", "88", "N/A", "140", " 20 ",
    "9", "12", "", "45", "182", "96"
]
```

```python
def clean_machakos(raw_data):
    valid_values = []
    invalid_months = []

    for i, value in enumerate(raw_data):
        try:
            valid_values.append(int(value))
        except ValueError:
            invalid_months.append(months[i])

    average = round(np.mean(valid_values))

    cleaned_values = []

    for i, value in enumerate(raw_data):
        try:
            cleaned_values.append(int(value))
        except ValueError:
            cleaned_values.append(average)

    print("Imputed months:", invalid_months)
    print("Imputed value:", average, "mm")

    return cleaned_values


machakos_clean = clean_machakos(machakos_raw)

print("Cleaned Machakos data:", machakos_clean)
```

```python
rainfall = np.array(clean_rows + [machakos_clean])

print("Combined rainfall array:")
print(rainfall)

print("\nArray shape:", rainfall.shape)
print("Array data type:", rainfall.dtype)
```

```python
annual_totals = rainfall.sum(axis=1)

monthly_averages = rainfall.mean(axis=0)

print("Annual rainfall totals per county:")
for county, total in zip(counties, annual_totals):
    print(f"{county}: {total} mm")

print("\nAverage rainfall per month across all counties:")
for month, average in zip(months, monthly_averages):
    print(f"{month}: {average:.2f} mm")
```

```python
wettest_indices = np.argmax(rainfall, axis=1)

wettest_months = np.array(months)[wettest_indices]

print("Wettest month per county:")

for county, month in zip(counties, wettest_months):
    print(f"{county}: {month}")
```

```python
dry_mask = rainfall < 50

dry_month_counts = dry_mask.sum(axis=1)

print("Number of dry months per county:")

for county, count in zip(counties, dry_month_counts):
    print(f"{county}: {count} dry months")
```

```python
# Calculate rainfall for consecutive months
consecutive_totals = rainfall[:, :-1] + rainfall[:, 1:]

# Find the index of the highest consecutive rainfall
planting_indices = np.argmax(consecutive_totals, axis=1)

# Get the corresponding planting windows
planting_windows = [
    f"{months[i]}–{months[i+1]}"
    for i in planting_indices
]

print("Recommended planting windows:")

for county, window in zip(counties, planting_windows):
    print(f"{county}: {window}")
```

```python
with open("rainfall_summary.csv", "w", newline="") as file:
    writer = csv.writer(file)

    # Write the column headings
    writer.writerow([
        "county",
        "annual_total_mm",
        "wettest_month",
        "dry_months",
        "planting_window"
    ])

    # Write the results for each county
    for row in zip(
        counties,
        annual_totals,
        wettest_months,
        dry_month_counts,
        planting_windows
    ):
        writer.writerow(row)

print("rainfall_summary.csv created successfully!")
```

## Planting Advisory

Based on the rainfall analysis, Nairobi, Nakuru, Kisumu and Mombasa have their highest combined rainfall in April and May, making this the recommended rainfall-based planting window for these counties. Farmers in Machakos should consider November and December, which recorded the highest combined rainfall for that county. Machakos requires particular attention because it recorded six dry months, the highest among the five counties, indicating a greater need for water conservation and careful soil-moisture management. Farmers should prepare their land before the recommended windows and monitor local weather forecasts before planting. These recommendations are based on historical monthly rainfall data and should be adjusted according to actual rainfall conditions, crop requirements and local agricultural advice.

## Data Quality Notes

The original Machakos dataset contained two invalid entries: "N/A" for April and an empty value for September. Both entries were replaced with 68 mm, calculated by rounding the average of the ten valid Machakos rainfall readings (67.5 mm). This imputation assumes that the missing values are reasonably represented by the average of the available readings. Consequently, the Machakos annual total, monthly averages and recommended planting window may be affected by the estimated values. The analysis uses the supplied monthly rainfall data and does not account for rainfall variability, soil conditions or future weather patterns.
